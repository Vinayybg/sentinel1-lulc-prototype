```python
import ee, json
from kaggle_secrets import UserSecretsClient

key = UserSecretsClient().get_secret("EE_KEY")
info = json.loads(key)
creds = ee.ServiceAccountCredentials(info["client_email"], key_data=key)
ee.Initialize(creds, project="lulc-sar-project")
print("Earth Engine ready, account:", info["client_email"])
```

```python
import ee
ids = {632:(-76.986,64.483), 10:(129.730,62.030), 70:(123.168,46.902), 333:(42.760,57.003), 526:(16.570,9.371)}
for pid,(lon,lat) in ids.items():
    pt = ee.Geometry.Point([lon,lat])
    base = (ee.ImageCollection("COPERNICUS/S1_GRD").filterBounds(pt)
            .filter(ee.Filter.eq("instrumentMode","IW"))
            .filter(ee.Filter.listContains("transmitterReceiverPolarisation","VV"))
            .filter(ee.Filter.listContains("transmitterReceiverPolarisation","VH")))
    for a,b in [("2025-01-01","2025-02-01"),("2025-02-01","2025-03-01"),("2025-03-01","2025-04-01")]:
        print(pid, a[:7], base.filterDate(a,b).size().getInfo())
```

```python
import os, ee
pt = ee.Geometry.Point([16.570, 9.371])
base = (ee.ImageCollection("COPERNICUS/S1_GRD").filterBounds(pt)
        .filter(ee.Filter.eq("instrumentMode","IW"))
        .filter(ee.Filter.listContains("transmitterReceiverPolarisation","VV"))
        .filter(ee.Filter.listContains("transmitterReceiverPolarisation","VH")))
for y in (2025, 2026):
    for m in range(1, 13):
        if y == 2026 and m > 9: break
        a = f"{y}-{m:02d}-01"
        b = f"{y+1}-01-01" if m == 12 else f"{y}-{m+1:02d}-01"
        print(f"{y}-{m:02d}", base.filterDate(a, b).size().getInfo())
```
