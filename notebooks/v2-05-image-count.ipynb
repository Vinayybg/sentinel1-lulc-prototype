```python
import glob, os, numpy as np, pandas as pd
PL = glob.glob("/kaggle/input/**/v2_places.csv", recursive=True)[0]
IN = os.path.dirname(PL); CONS = os.path.join(IN, "cons")
pl = pd.read_csv(PL)
print(pl.columns.tolist()); print(pl.head(3).to_string())
n_patches = {}
for pid in pl.id:
    f = f"{CONS}/{int(pid):04d}.npz"
    n_patches[pid] = np.load(f)["cons"].shape[0] if os.path.exists(f) else 0
pl["patches"] = pl.id.map(n_patches)
pl["images"] = pl.patches * pl.months_ok
print("\npatches per place:", pl.patches.describe().round(1).to_dict())
print("months_ok per place:", pl.months_ok.describe().round(1).to_dict())
print("images per place (patches x valid months):", pl.images.describe().round(1).to_dict())
print("\nplaces with >=100 images:", int((pl.images >= 100).sum()), "of", len(pl))
print("by split:\n", pl.groupby("split").images.agg(["count", "min", "median", "max"]).to_string())
low = pl[pl.images < 100]
print("\nplaces below 100:", len(low)); print(low[["id", "split", "months_ok", "patches", "images"]].head(40).to_string())
```
